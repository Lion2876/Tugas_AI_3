```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('insurance.csv')
```

```python
print(df.head())
```

```python
x = df['bmi'].values
y = df['charges'].values
```

```python
w = 0
b = 0
lr = 0.0001
epochs = 100
```

```python
for epoch in range(epochs):
    for i in range(len(x)):
        y_pred = w * x[i] + b
        error = y_pred - y[i]

        w = w - lr * 2 * error * x[i]
        b = b - lr * 2 * error
```

```python
print("Weight:", w)
print("Bias:", b)
```

```python
plt.scatter(x, y)
plt.plot(x, w*x + b)
plt.xlabel("BMI")
plt.ylabel("Charges")
plt.title("Linear Regression")
plt.show()
```
